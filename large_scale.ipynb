## 1. Start with one C file

We start with a tiny calculator split into modules. The important pattern is:

- the header declares the interface;
- the `.c` file implements it;
- the caller includes the header;
- all required object files are linked into the executable.


````python
from pathlib import Path

WORKDIR = Path.cwd() / "large_scale_lecture_project"
if WORKDIR.exists():
    import shutil
    shutil.rmtree(WORKDIR)
WORKDIR.mkdir(exist_ok=True)

hello = WORKDIR / "hello.c"
hello.write_text(r'''
#include <stdio.h>

int main(void)
{
    printf("Hello from C!\n");
    return 0;
}
'''.lstrip())

print(hello.read_text())
````

````python
result = subprocess.run(
    ["gcc", "-ansi", "-Wall", "-Wextra", "-Werror", "-pedantic-errors", "-o", str(WORKDIR / "hello"), str(hello)],
    text=True, capture_output=True
)
print(result.stdout)
print(result.stderr)
assert result.returncode == 0
````

````python
result = subprocess.run([str(WORKDIR / "hello")], text=True, capture_output=True)
print(result.stdout)
````

# 2. C modules: split the program

A large program should not become one enormous `.c` file.

We will build a tiny calculator with:
- `main.c` — user-facing program;
- `math_utils.c` / `math_utils.h` — arithmetic functions;
- `string_utils.c` / `string_utils.h` — formatting functions.

The `.h` files contain **declarations**; the `.c` files contain **definitions**.


````python
# Create the initial multi-file project
(WORKDIR / "src").mkdir(exist_ok=True)

(WORKDIR / "src" / "math_utils.h").write_text(r'''
#ifndef MATH_UTILS_H
#define MATH_UTILS_H

int add(int a, int b);
int multiply(int a, int b);
int subtract(int a, int b);

#endif
'''.lstrip())

(WORKDIR / "src" / "math_utils.c").write_text(r'''
#include "math_utils.h"

int add(int a, int b)
{
    return a + b;
}

int multiply(int a, int b)
{
    return a * b;
}

int subtract(int a, int b)
{
    return a - b;
}
'''.lstrip())

(WORKDIR / "src" / "string_utils.h").write_text(r'''
#ifndef STRING_UTILS_H
#define STRING_UTILS_H

void print_result(const char *operation, int value);

#endif
'''.lstrip())

(WORKDIR / "src" / "string_utils.c").write_text(r'''
#include <stdio.h>
#include "string_utils.h"

void print_result(const char *operation, int value)
{
    printf("%s = %d\n", operation, value);
}
'''.lstrip())

(WORKDIR / "src" / "main.c").write_text(r'''
#include "math_utils.h"
#include "string_utils.h"

int main(void)
{
    print_result("2 + 3", add(2, 3));
    print_result("4 * 5", multiply(4, 5));
    print_result("4 - 5", subtract(4, 2));
    return 0;
}
'''.lstrip())

print("\n".join(p.name for p in sorted((WORKDIR / "src").iterdir())))
````

````python
# Compile and link all three source files in one command.
result = subprocess.run(
    ["gcc", "-ansi", "-Wall", "-Wextra", "-pedantic-errors",
     "-o", str(WORKDIR / "calculator"),
     str(WORKDIR / "src" / "main.c"),
     str(WORKDIR / "src" / "math_utils.c"),
     str(WORKDIR / "src" / "string_utils.c")],
    text=True, capture_output=True
)
print(result.stderr)
assert result.returncode == 0

run = subprocess.run([str(WORKDIR / "calculator")], text=True, capture_output=True)
print(run.stdout)
````

## 2.1 Compile → link: see the intermediate files

The previous command hides an important process. Compilation and linking are separate steps.

Let's create object files explicitly.


````python
for source in ["main.c", "math_utils.c", "string_utils.c"]:
    result = subprocess.run(
        ["gcc", "-ansi", "-Wall", "-Wextra", "-Werror", "-pedantic-errors", "-c", str(WORKDIR / "src" / source),
         "-o", str(WORKDIR / "src" / source.replace(".c", ".o"))],
        text=True, capture_output=True
    )
    print(source, "->", result.returncode)
    if result.stderr:
        print(result.stderr)

print("Object files:", [p.name for p in (WORKDIR / "src").glob("*.o")])
````

````python
objects = [str(p) for p in sorted((WORKDIR / "src").glob("*.o"))]
result = subprocess.run(
    ["gcc", "-o", str(WORKDIR / "calculator_from_objects"), *objects],
    text=True, capture_output=True
)
print(result.stderr)
assert result.returncode == 0
print(subprocess.run([str(WORKDIR / "calculator_from_objects")], text=True, capture_output=True).stdout)
````

# 4. Make: let the computer manage the dependency graph


## What is a Makefile?

A **Makefile is a build recipe + dependency map**. Instead of remembering which `gcc` commands to run,
we describe **what depends on what**, and `make` decides what needs to be rebuilt.

```text
                  DEPENDENCIES                         ACTION
              ┌─────────────────┐
              │   main.c        │
              └────────┬────────┘
                       │ compile
                       ▼
                    main.o ──────────┐
                                     │
              ┌─────────────────┐    │
              │ math_utils.c    │    │
              └────────┬────────┘    │
                       │ compile     │
                       ▼             │
                  math_utils.o ──────┤
                                     │ link
              ┌─────────────────┐    │
              │ string_utils.c  │    │
              └────────┬────────┘    │
                       │ compile     │
                       ▼             │
                string_utils.o ──────┘
                                     │
                                     ▼
                               ┌────────────┐
                               │ calculator │
                               └────────────┘
```

The core Makefile pattern is simply:

```make
target: prerequisites
	command to build target
```

For example:

```make
math_utils.o: src/math_utils.c src/math_utils.h
	gcc -ansi -Wall -Wextra -Werror -pedantic-errors -c src/math_utils.c -o math_utils.o
```

**The useful part:** if only `math_utils.c` changes, Make does not need to rebuild everything.
It follows the dependency graph and rebuilds only what is affected.



```shell
CC = gcc
CFLAGS = -ansi -Wall -Wextra -Werror -pedantic-errors

calculator: main.o math_utils.o string_utils.o
	$(CC) -o calculator main.o math_utils.o string_utils.o

main.o: src/main.c src/math_utils.h src/string_utils.h
	$(CC) $(CFLAGS) -c src/main.c

math_utils.o: src/math_utils.c src/math_utils.h
	$(CC) $(CFLAGS) -c src/math_utils.c

string_utils.o: src/string_utils.c src/string_utils.h
	$(CC) $(CFLAGS) -c src/string_utils.c

clean:
	rm -f calculator *.o
```


````python
makefile = WORKDIR / "Makefile"
makefile.write_text(r'''
CC = gcc
CFLAGS = -ansi -Wall -Wextra -Werror -pedantic-errors
OBJECTS = main.o math_utils.o string_utils.o

calculator: $(OBJECTS)
	$(CC) -o calculator $(OBJECTS)

main.o: src/main.c src/math_utils.h src/string_utils.h
	$(CC) $(CFLAGS) -c src/main.c

math_utils.o: src/math_utils.c src/math_utils.h
	$(CC) $(CFLAGS) -c src/math_utils.c

string_utils.o: src/string_utils.c src/string_utils.h
	$(CC) $(CFLAGS) -c src/string_utils.c

clean:
	rm -f calculator *.o
'''.lstrip())

````

````python
def run_make(*args):
    result = subprocess.run(["make", *args], cwd=WORKDIR, text=True, capture_output=True)
    print(result.stdout, end="")
    if result.stderr:
        print(result.stderr, end="")
    return result

result = run_make()
assert result.returncode == 0
````

# 5. Git and Branches

## What is Git?

**Git is a version control system.**

It keeps a history of changes to your project, allowing you to:

- see what changed and when
- go back to an earlier version
- work on different features independently
- collaborate with other developers
- combine changes made by different people

Think of Git as a **timeline of your project**:

```text
Project history

   A              B              C              D
   ●──────────────●──────────────●──────────────●
   │              │              │              │
 initial       add module     fix bug       add tests
 commit        commit         commit         commit

```
Each point is a **commit**.

A commit is a snapshot of the project at a particular point in time, together with information about what changed.


### The basic Git workflow


When you modify files, Git lets you choose which changes should become part of the next commit.

```text
                    git add                 git commit
                       │                        │
                       ▼                        ▼
 Working directory → Staging area → Repository history
    (changes)     (selected changes)       (snapshot)

```
For example:

```shell
git status
git add src/math_utils.c
git commit -m "Fix multiplication"
```

# 6. Branches: work on a feature without disturbing `main`


###What is a branch?
A branch is an independent line of development.



 Imagine that your project currently looks like this:

 ```text
                    main
                     │
                     ▼
A ──────── B ─────── C
```

You want to add a new feature, but you don't want to modify main directly.
Create a branch:

 ```text
                    feature
                       │
                       ▼
A ──────── B ─────── C
                       │
                    main

```

Now you can make commits on the feature branch:

```text
                    feature
                       │
                       ▼
A ──────── B ─────── C ─────── D ─────── E
                       │
                      main

```

`main` still points to `C`, while your feature branch contains the new work.

````python
result = subprocess.run(
    ["git", "checkout", "-b", "feature_division"],
    cwd=WORKDIR, text=True, capture_output=True
)
print(result.stdout, result.stderr)
````

````python
# Add subtraction to the module.
h = WORKDIR / "src" / "math_utils.h"
c = WORKDIR / "src" / "math_utils.c"
main = WORKDIR / "src" / "main.c"

h.write_text(h.read_text().replace("int multiply(int a, int b);", "int multiply(int a, int b);\nint divide(int a, int b);"))
c.write_text(c.read_text() + "\nint divide(int a, int b)\n{\n    return a / (b + 1e-12);\n}\n")
main.write_text(main.read_text().replace(
    'print_result("10 + 7", add(10, 7));',
    'print_result("10 + 7", add(10, 7));\n    print_result("10 / 7", divide(10, 7));'
))

run_make()
````

````python
result = subprocess.run(["git", "commit", "-am", "Add division operation"],
                        cwd=WORKDIR, text=True, capture_output=True)
print(result.stdout)
````

````python
# Publish the feature branch to the local remote.
result = subprocess.run(
    ["git", "push", "-u", "origin", "feature_division"],
    cwd=WORKDIR, text=True, capture_output=True
)
print(result.stdout)
print(result.stderr)
assert result.returncode == 0
````

We work on the new branch:
```text
                         feature_division
                              │
                              ▼
A ─────── B ─────── C ─────── D ─────── E
                     ▲
                     │
                    main
```
We can experiment, make mistakes, and add commits without affecting `main`.

### Merge it




Once the feature is ready, we can merge the branch back into `main`:
```text
                       D ─────── E
                      /           \
A ─────── B ─────── C               F
                     \             /
                       ───────────
                           main
 ```

The merge combines the work from the two lines of development.                          

````python
for command in [
    ["git", "checkout", "main"],
    ["git", "pull", "origin", "main"],
    ["git", "checkout", "feature_division"],
    ["git", "merge", "main", "-m", "Merged main"],
]:
    result = subprocess.run(command, cwd=WORKDIR, text=True, capture_output=True)
    print("$", " ".join(command))
    print(result.stdout)
    print(result.stderr)
    if result.returncode != 0:
        raise RuntimeError("Command failed: " + " ".join(command))

print(subprocess.run(
    ["git", "log", "--graph", "--oneline", "--all"],
    cwd=WORKDIR, text=True, capture_output=True, check=True
).stdout)
````

# 7. A controlled merge conflict

Conflicts are easier to understand if we create one intentionally.

Two branches will make incompatible edits to the same line. Your task is to inspect the conflict and resolve it.


````python
# Create feature_format and commit its formatting change.
subprocess.run(["git", "checkout", "-b", "feature_format"], cwd=WORKDIR, check=True)

path = WORKDIR / "src" / "string_utils.c"
text = path.read_text().replace(
    'printf("%s = %d\\n", operation, value);',
    'printf("[RESULT] %s = %d\\n", operation, value);',
)
path.write_text(text)
subprocess.run(["git", "commit", "-am", "Add result prefix"], cwd=WORKDIR, check=True)

# Switch to main and commit a conflicting formatting change.
subprocess.run(["git", "checkout", "main"], cwd=WORKDIR, check=True)

path = WORKDIR / "src" / "string_utils.c"
text = path.read_text().replace(
    'printf("%s = %d\\n", operation, value);',
    'printf("RESULT: %s = %d\\n", operation, value);',
)
path.write_text(text)
subprocess.run(["git", "commit", "-am", "Change result formatting"], cwd=WORKDIR, check=True)

# Return to feature_format and merge main into it.
subprocess.run(["git", "checkout", "feature_format"], cwd=WORKDIR, check=True)
result = subprocess.run(
    ["git", "merge", "main"],
    cwd=WORKDIR, text=True, capture_output=True,
)
print(result.stdout)
print(result.stderr)
````

One simple technique in C is to compile a dedicated test executable that links the module being tested with a test `main`.

# 10. Integration testing

Unit tests check components. Integration tests check that components work together.

Here, the integration boundary is:

```text
main → math_utils → result → string_utils → stdout
```

# 11. Performance testing: measure before optimizing

Performance testing asks a quantitative question.

Instead of saying “this feels slow”, measure elapsed time. For a tiny program the numbers are not interesting, but the workflow is.


````python
import time

start = time.perf_counter()
for _ in range(100):
    subprocess.run([str(WORKDIR / "calculator")], stdout=subprocess.DEVNULL, check=True)
elapsed = time.perf_counter() - start

print(f"100 executions: {elapsed:.4f} seconds")
print(f"Average: {elapsed / 100:.6f} seconds")
````

# 12. Logging

`printf()` is useful for user-facing output. Logging is useful for understanding what a program is doing while it runs.

Typical levels include:
- `DEBUG` — detailed diagnostic information;
- `INFO` — normal operational events;
- `WARNING` — something unexpected but recoverable;
- `ERROR` — an operation failed;
- `CRITICAL` — severe failure.

Let's first use Python's logging module to see the idea.


````python
import logging

logger = logging.getLogger("lecture")
logger.setLevel(logging.DEBUG)

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(logging.Formatter("%(levelname)s: %(message)s"))
    logger.addHandler(handler)

logger.debug("Detailed state")
logger.info("Program started")
logger.warning("Input is unusual")
logger.error("An operation failed")
````

## 12.1 Add logging to the C program

C does not have a standard logging framework comparable to Python's `logging`, but a project can define a small logging interface.

We will add a minimal logger that writes levels and messages to `stderr`.


````python
(WORKDIR / "src" / "logger.h").write_text(r'''
#ifndef LOGGER_H
#define LOGGER_H

void log_info(const char *message);
void log_warning(const char *message);
void log_error(const char *message);

#endif
'''.lstrip())

(WORKDIR / "src" / "logger.c").write_text(r'''
#include <stdio.h>
#include "logger.h"

void log_info(const char *message)
{
    fprintf(stderr, "INFO: %s\n", message);
}

void log_warning(const char *message)
{
    fprintf(stderr, "WARNING: %s\n", message);
}

void log_error(const char *message)
{
    fprintf(stderr, "ERROR: %s\n", message);
}
'''.lstrip())

# Add a simple startup log to main.c.
path = WORKDIR / "src" / "main.c"
text = path.read_text()
text = text.replace('#include "math_utils.h"', '#include "math_utils.h"\n#include "logger.h"')
text = text.replace('int main(void)\n{', 'int main(void)\n{\n    log_info("calculator started");')
path.write_text(text)

# Add logger.o to the automatic Makefile.
make_text = makefile.read_text()
make_text = make_text.replace(
    "OBJECTS = main.o math_utils.o string_utils.o",
    "OBJECTS = main.o math_utils.o string_utils.o logger.o"
)
make_text = make_text.replace(
    "clean:",
    "logger.o: src/logger.c src/logger.h\n\t$(CC) $(CFLAGS) -c src/logger.c\n\nclean:"
)
makefile.write_text(make_text)

run_make()
````

# 13. CI/CD: From Code to Release

![](https://www.testingdocs.com/wp-content/uploads/CiCD-pipeline.png)

Imagine you change one line of code. How do you know you haven't broken something else? And how does that change reach users?

**CI/CD automates the journey from a code change to a release.**

## Three stages

### 1. Continuous Integration (CI) — Does it work?

Every proposed change triggers a build and automated tests. Failures are caught early.

    Code change → Build → Tests → Pass / Fail

### 2. Continuous Delivery — Is it ready to release?

Every change that passes CI is prepared for release. The software can be released at any time, but a person must approve the deployment.

    CI passes → Release-ready → Manual approval → Deploy

### 3. Continuous Deployment — Release it automatically!

Every change that passes all required checks is automatically deployed to production, without manual release approval.

    CI passes → Release-ready → Automatic deployment

**The key difference:** Continuous *delivery* keeps software ready to release; continuous *deployment* automatically releases it to production.

# 16. Connecting to a real remote repository

Today, `origin` points to a **local bare repository** so that `push` and `pull` work without authentication. To connect a project to a real remote such as GitHub, create a repository on the hosting service, then configure its URL.

For an existing local repository, use the HTTPS URL shown by GitHub:

```bash
git remote add origin https://github.com/USERNAME/REPOSITORY.git
git push -u origin main
```

If `origin` already exists, update it instead:

```bash
git remote set-url origin https://github.com/USERNAME/REPOSITORY.git
```

Useful checks:

```bash
git remote -v
git branch -vv
```

**Authentication note:** setting a remote URL does not grant permission to push. GitHub requires authentication (for example, via Git Credential Manager, GitHub CLI, or a personal access token). In Colab, Git cannot reliably prompt for credentials inside a notebook output cell, so complete authentication separately before pushing. Never put a personal access token directly in a shared notebook.
